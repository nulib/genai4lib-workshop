## Visualize the data

```python
import json
import numpy as np
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import mplcursors
import textwrap

# Read and parse JSON file
with open('./pg_embeddings_export.json', 'r') as f:
    data = json.load(f)

# Extract embeddings and documents
embeddings = []
documents = []
for item in data:
    # Convert the string of the embedding to a Python list
    embedding = json.loads(item['embedding'].replace("'", '"'))
    embeddings.append(embedding)
    documents.append(item['document'])

embeddings_array = np.array(embeddings)

# Perform t-SNE dimensionality reduction
tsne = TSNE(n_components=2, perplexity=4, random_state=42)
two_d = tsne.fit_transform(embeddings_array)

# Create the scatter plot
plt.figure(figsize=(10, 10))
scatter = plt.scatter(
    two_d[:, 0],
    two_d[:, 1],
    c=np.arange(len(documents)),
    cmap='viridis',
    marker='o'
)

plt.title('t-SNE Visualization of Document Embeddings')
plt.xlabel('t-SNE Feature 1')
plt.ylabel('t-SNE Feature 2')
plt.grid(True)

def on_add_annotation(sel):
    """
    This function is called whenever the user hovers over a point.
    We retrieve the document text at sel.index and set it as the annotation.
    """
    # Get the raw document text
    doc_text = documents[sel.index]
    # wrap text to avoid overly wide tooltip
    wrapped_text = "\n".join(textwrap.wrap(doc_text, width=70))
    
    sel.annotation.set_text(wrapped_text)
    sel.annotation.get_bbox_patch().set(fc="white", alpha=0.9)
    sel.annotation.set_fontsize(9)

# Attach the hover cursor/annotation
cursor = mplcursors.cursor(scatter, hover=True)
cursor.connect("add", on_add_annotation)

plt.tight_layout()
plt.show()
```
