# Agents

## Tool-calling agent

```python
from smolagents import CodeAgent, DuckDuckGoSearchTool, LiteLLMModel

model = LiteLLMModel(model_id="bedrock/amazon.nova-lite-v1:0")
agent = CodeAgent(tools=[DuckDuckGoSearchTool()], model=model)

agent.run("What time is does the cafe at open in the Northwestern University Library?")
```
