# Multimodal Language Models

```python
# Helper functions

def encode_image(image_path):
    import base64

    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")

def download_and_encode_image(url):
    import requests
    import base64

    response = requests.get(url)
    response.raise_for_status()
    return base64.b64encode(response.content).decode('utf-8')
```

### Reading an image from a file path

```python
from litellm import completion

image_path = "../../images/headings/01_embeddings_01_01_generating_embeddings.png"

# Getting the base64 string
base64_image = encode_image(image_path)

resp = completion(
    model="bedrock/amazon.nova-lite-v1:0",
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "Whats in this image?"},
                {
                    "type": "image_url",
                    "image_url": {
                        "url": "data:image/png;base64," + base64_image
                    },
                },
            ],
        }
    ],
)

print(f"\nResponse: {resp.choices[0].message.content}")
```

## Reading an image from a URL

```python
from IPython.display import Image

Image(url="https://repositorio.colmex.mx/downloads/w95052629")
```

```python
try:
    base64_image = download_and_encode_image("https://repositorio.colmex.mx/downloads/w95052629")
    
    resp = completion(
        model="bedrock/amazon.nova-lite-v1:0",
        messages=[
            {
                "role": "user",
                "content": [
                    {"type": "text", "text": "Whats in this image?"},
                    {
                        "type": "image_url",
                        "image_url": {
                            "url": f"data:image/jpeg;base64,{base64_image}"
                        },
                    },
                ],
            }
        ],
    )
    
    print(f"\nResponse: {resp.choices[0].message.content}")

except requests.RequestException as e:
    print(f"Error downloading image: {e}")
```
